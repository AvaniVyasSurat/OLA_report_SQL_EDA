## Total Trips

###
SELECT COUNT(DISTINCT tripid) FROM trip_details;

### **Explanation:**
- This query calculates the **Total Trips** requested by the users.
- It works by counting the unique `tripid` values in the `trip_details` table using `COUNT(DISTINCT tripid)`. This ensures each trip is counted only once, even if it has multiple events recorded in the system.

## Total Drivers

###
SELECT COUNT(DISTINCT driverid) FROM trips;

## Which 2 locations had the most trips ?

###
SELECT loc_from,loc_to,COUNT(DISTINCT tripid) FROM trips
GROUP BY loc_from, loc_to
ORDER BY COUNT(DISTINCT tripid) DESC;

### **Explanation:**
- This query finds the **top two location pairs** (from-to) with the highest number of completed trips.
- It works by grouping the trips by both starting location (`loc_from`) and destination (`loc_to`), counting the unique trips (`COUNT(DISTINCT tripid)`) for each combination, and sorting the list in descending order to place the most frequent routes at the top.

## Which Duration had more trips ?

###
SELECT *, rank() over(ORDER BY cnt DESC) rnk FROM
(SELECT duration, COUNT(DISTINCT tripid) cnt FROM trips
GROUP BY duration)b;

### **Explanation:**
- This query ranks different **trip durations** based on the total number of trips they had.
- It works using a subquery (called `b`) that groups trips by `duration` and counts the unique trips (`COUNT(DISTINCT tripid) as cnt`). The outer query then applies the `rank() over (order by cnt desc)` window function to assign a numerical rank to each duration based on demand.

## Which single duration had the absolute highest number of trips?

###
SELECT * FROM
(SELECT *, rank() over(ORDER BY cnt DESC) rnk FROM
(SELECT duration, COUNT(DISTINCT tripid) cnt FROM trips
GROUP BY duration)b)c
WHERE rnk=1;

### **Explanation:**
- This query filters the ranked duration data to return **only the single duration value** with the absolute highest number of trips.
- It works by wrapping the previous ranking query inside another outer subquery (called `c`), which allows us to use the `WHERE rnk = 1` filter to isolate the most popular trip duration.

## Which area got the highest fares, cancellations ?

###
SELECT * FROM (SELECT*,rank() over(ORDER BY fare DESC) rnk
FROM
(SELECT loc_from, SUM(fare) fare FROM trips
GROUP BY loc_from)b)c
WHERE rnk=1;

### **Explanation:**
- This query finds the **location area** that generated the highest total fare amount.
- It works by grouping the trips by pickup location (`loc_from`) and summing up the fares (`SUM(fare)`). It then uses `rank() over(order by fare desc)` within nested subqueries to identify and filter for the location ranked as number 1 (`rnk = 1`).

## Which area got the highest cancellations from the driver side?

###
SELECT * FROM (SELECT *, rank() over(ORDER BY can desc) rnk
FROM
(SELECT loc_from, COUNT(*)-SUM(driver_not_cancelled) can
FROM trip_details
GROUP BY loc_from)b)c
WHERE rnk=1;

### **Explanation:**
- This query identifies the **location area** with the highest number of driver-side cancellations.
- It works by calculating driver cancellations per pickup location (`COUNT(*) - SUM(driver_not_cancelled)`), ranking those cancellation counts in descending order, and filtering for the top rank (`rnk = 1`).

## Which area got the highest cancellations from the customer side?

###
SELECT * FROM (SELECT *,rank() over(ORDER BY can DESC) rnk
FROM
(
SELECT loc_from, COUNT(*)-sum(customer_not_cancelled) can
FROM trip_details
GROUP BY loc_from)b)c
WHERE rnk=1;

### **Explanation:**
- This query identifies the **location area** with the highest number of customer-side cancellations.
- Similar to the driver cancellation query, it calculates customer cancellations per location using `COUNT(*) - SUM(customer_not_cancelled)`. It ranks these cancellation counts in descending order and filters using `WHERE rnk = 1` to find the most cancelled pickup location.
